```python
import numpy as np
import matplotlib.pyplot as plt

import utils
```

```python
def plot_volume_voxels(volume):
    
    mu = np.mean(volume)
    vol = np.where(volume > mu, volume, 0)
    filled = volume > mu
    vol_norm = (vol - vol.min()) / (vol.max() - vol.min())
    
    cmap = plt.cm.viridis
    colors = cmap(vol_norm)
    
    facecolors = np.zeros_like(colors)
    facecolors[filled] = colors[filled]
    
    fig = plt.figure()
    ax = fig.add_subplot(projection='3d')
    ax.voxels(filled, facecolors=facecolors, alpha=0.4, edgecolor='none')
    
    # ax.view_init(elev=0, azim=45, roll=0)
    
    # ax.set_xlabel('$X$', fontsize=10)
    # ax.set_ylabel('$Y$', fontsize=10)
    # ax.set_zlabel('$Z$', fontsize=10)
    
    plt.show()
```

```python
L = 32
sigma = 2

V = np.zeros((L,L,L))

V[16, 16, 16] = 1
V[16, 8, 12] = 1
V[16, 8, 20] = 1

g = utils.generate_centered_gaussian(L=L, d=3, sigma=sigma)

V = utils.signal_convolution(V, g)
```

```python
plot_volume_voxels(V)
```
